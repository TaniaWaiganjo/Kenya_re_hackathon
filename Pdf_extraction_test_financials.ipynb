```python
!pip install PyMuPDF

import fitz  # PyMuPDF
import os

def is_text_based(pdf_path, min_text_threshold=10, sample_pages=3):
    """
    Check if a PDF is text-based or scanned (image-based).

    Args:
        pdf_path (str): Path to the PDF file.
        min_text_threshold (int): Minimum number of characters to consider a page as "text-based".
        sample_pages (int): Number of pages to sample (default: 3).

    Returns:
        bool: True if PDF is text-based, False if scanned or image-based.
    """
    doc = fitz.open(pdf_path)
    num_pages = len(doc)
    pages_to_check = min(sample_pages, num_pages)
    text_pages = 0

    for i in range(pages_to_check):
        text = doc[i].get_text().strip()
        if len(text) >= min_text_threshold:
            text_pages += 1

    doc.close()

    return text_pages >= pages_to_check // 2 + 1
```

```python
# Create a list of the PDF files you want to check
pdf_files_to_check = [
    Fekan2022,
    Fekan2023,
    AvidFinancials,
    DirectOFinancials,
    Sentimental2022,
    Sentimental2020
]

# Iterate through the list and check each file
for pdf_file in pdf_files_to_check:
    if is_text_based(pdf_file):
        print(f"✅ {pdf_file} is text-based. Proceed with PyMuPDF.")
    else:
        print(f"🖼️ {pdf_file} appears scanned or image-based. Use OCR (e.g., Tesseract).")
```

```python

def extract_text_from_pdf(pdf_path):
    """
    Extracts all text from a PDF file using PyMuPDF.

    Args:
        pdf_path (str): Path to the PDF file.

    Returns:
        str: Extracted text content.
    """
    try:
        doc = fitz.open(pdf_path)
        text = ""
        for page in doc:
            text += page.get_text()
        doc.close()
        return text
    except Exception as e:
        return f"Error extracting text from {pdf_path}: {e}"
```
